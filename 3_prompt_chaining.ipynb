```python
from langgraph.graph import StateGraph,START,END
from dotenv import load_dotenv
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
load_dotenv()
model= ChatGoogleGenerativeAI(model='gemini-2.5-flash')
```

```python
class BlogState(TypedDict):
    title:str
    outline:str
    content:str
```

```python
def create_outline(state:BlogState)->BlogState:
    title=state['title']
    prompt=f'generete detail outline for a blog on the topic - {title}'
    result= model.invoke(prompt).text
    state['outline']=result
    return state

def create_blog(state:BlogState)-> BlogState:
     title= state['title']
     outline=state['outline']
     prompt = f'Write a detailed blog on the title - {title} using the follwing outline \n {outline}'
     result= model.invoke(prompt).text
     state['content']= result
     return state
```

```python
graph= StateGraph(BlogState)
graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)

graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog',END)

work_flow=graph.compile()
```

```python
final_result=work_flow.invoke({'title':'Rise of AI in India'})

print(final_result['outline'])
```
